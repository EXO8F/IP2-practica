```python
##uitwerking van alles behalve met errors meegenomen in de curve fit.



import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
from decimal import Decimal




## metingen in arrays##
U_meas_3 = np.array([0.12,0.19,0.29,2.28,0.99,1.29,1.59,2.98,4.98,3.98,5.98,6.98,7.97,14.96,0.015,0.065,0.39,0.49,0.59,0.69,0.245,0.295,0.344,0.444,0.544,0.943,0.843])
U_meas_5 = np.array([0.606*10**-3,3.665*10**-3,16.745*10**-3,1.57475,0.47799,0.72494,0.97659,2.17935,3.9213,3.0484,4.7959,5.6725,6.5498,12.7066,0.070*10**-3,0.321*10**-3,51.171*10**-3,105.400*10**-3,171.101*10**-3,0.24346,8.180*10**-3,16.819*10**-3,30.995*10**-3,76.268*10**-3,137.216*10**-3,0.43777,0.35838])

## Weerstanden gemeten met 5 1/2 dmm ##
R1 = 33000 #ohm
R2 = 998.70 #ohm
R3 = 9975.5 #ohm

## constantes ##
q = 1.602176634*(10**(-19))
Kb= 1.380649*(10**(-23))
T = 293.15
I_R1 = U_meas_5/R1
I_Diode = I_R1
#n=2




## error of measurements
D_R2= (0.020*R2 + 0.003*200)/100
D_R1=(0.020*R3 + 0.010*200000)/100
D_R3=(0.020*R1 + 0.003*20000)/100


U_meas_5_errors = np.array([])
for i in range(len(U_meas_5)):
    if U_meas_5[i] < 2:
        U_meas_5_errors = np.append(U_meas_5_errors,(0.015*U_meas_5[i] + 0.004*0.2)/100)
    else:
        U_meas_5_errors = np.append(U_meas_5_errors,(0.015*U_meas_5[i] + 0.003*2)/100)


U_meas_3_errors = np.array([])
for i in range(len(U_meas_3)):
    num = str(U_meas_3[i])
    decimals = Decimal(num).as_tuple().exponent
    U_meas_3_errors = np.append(U_meas_3_errors,((0.7*U_meas_3[i])/100)+1*(10**decimals))

#error in I_D
I_D_error = np.sqrt(((U_meas_5_errors/U_meas_5)**2)+(D_R1/R1)**2)


#helse berekening U_D_error  =
k = R3 / (R2 + R3)    #factor van hoeveel stroom er door R2 gaat voor R3
UR2_zonder_R3= U_meas_3 - R2 * U_meas_5 / R1

#partiele afgeleides 
dU_dU3 = k
dU_dU5 = -(k * R2 / R1 + 1)
dU_dR1 = k * R2 * U_meas_5 / R1**2
dU_dR2 = -k / (R2 + R3) * UR2_zonder_R3 - k * U_meas_5 / R1
dU_dR3 = R2 / (R2 + R3)**2 * UR2_zonder_R3

U_D_error = np.sqrt((dU_dU3 * U_meas_3_errors)**2 +(dU_dU5 *U_meas_5_errors)**2 +(dU_dR1 * D_R1)**2 +(dU_dR2 * D_R2)**2 +(dU_dR3 * D_R3)**2)




## Berekeningen van U_diode en I_diode ##
I_D = U_meas_5 / R1
V_ondertak = (U_meas_3 - R2 * I_D) * R3 / (R2 + R3)
U_Diode = V_ondertak - U_meas_5

#formule Id: I0*(e**((-q*U_Diode)/(n*Kb*T)))
#e macht dus kunnen we een ln plot transformatie maken
# ln(I_D) = Ln(I0) + (q*U_diode)/(n*Kb*T)-1

## curve fit voor ln functie I_D ##
def ln_ID(U_Diode, n, I0):
    return np.log(I0)+np.log(np.exp((q*U_Diode)/(n*Kb*T))-1)

popt, pcov = curve_fit(ln_ID,U_Diode,np.log(I_Diode),p0=(2,1e-10))
n, I0 = popt
print(n,I0)

##
U_linspace = np.linspace(min(U_Diode),max(U_Diode),101)
plt.scatter(U_Diode,np.log(I_Diode))
plt.plot(U_linspace,ln_ID(U_linspace,*popt))
```

```python
##curve fit met errors ik snap het niet meer


mask  = (I_Diode > 0) & (U_Diode > 0.01)
x     = U_Diode[mask]
y     = np.log(I_Diode[mask])
sig_x = U_D_error[mask]
sig_y = I_D_error[mask]

def model(U, n, lnI0):
    return lnI0 + np.log(np.expm1(q * U / (n * Kb * T)))


p       = np.array([2.0, np.log(1e-10)])
sig_eff = sig_y.copy()

for i in range(20):
    p_old = p.copy()
    p, pcov = curve_fit(model, x, y, p0=p, sigma=sig_eff, absolute_sigma=True)

    a     = q / (p[0] * Kb * T)
    slope = a / (1 - np.exp(-a * x))            # d(model)/dU, overflow-safe
    sig_eff = np.sqrt(sig_y**2 + (slope * sig_x)**2)

    if np.all(np.abs(p - p_old) < 1e-8 * np.abs(p)):
        break

n, lnI0 = p
n_err, lnI0_err = np.sqrt(np.diag(pcov))
I0     = np.exp(lnI0)
I0_err = I0 * lnI0_err

chi2_red = np.sum(((y - model(x, *p)) / sig_eff)**2) / (len(x) - 2)

print(f"iterations: {i+1}")
print(f"n    = {n:.3f} ± {n_err:.3f}")
print(f"I0   = {I0:.3e} ± {I0_err:.1e} A")
print(f"chi2_red = {chi2_red:.2f}")


U_s = np.linspace(x.min(), x.max(), 200)
plt.errorbar(U_Diode, I_Diode, xerr=U_D_error, yerr=I_Diode * I_D_error,
             fmt='.', label='measured')
plt.plot(U_s, np.exp(model(U_s, *p)), 'k--', label='fit')
plt.yscale('log')
plt.xlabel(r'$U_{diode}$ (V)')
plt.ylabel(r'$I$ (A)')
plt.legend()
plt.show()
```
